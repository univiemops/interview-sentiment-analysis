# Create files for LIWC analyses from raw transcripts

```python
import pandas as pd
from pathlib import Path
import re

HERE = Path.cwd()

# Folder containing the input Excel files
input_folder = HERE.parent / "transcripts"

# Output file
output_file = HERE / "merged_patient_texts.xlsx"

results = []

# Loop through all subject files
for file in sorted(input_folder.glob("T*.xlsx")):

    # Extract subject ID (e.g., T0001)
    match = re.match(r"^(T\d+)", file.stem)

    if not match:
        continue

    subject_id = match.group(1)

    # Read Excel file
    df = pd.read_excel(file)

    # Select only patient rows
    patient_df = df[
        df["role"].astype(str).str.strip().str.lower().str.rstrip(":") == "patient"
    ]

    # Merge German text
    german_text = " ".join(
        patient_df["text"].dropna().astype(str).str.strip()
    )

    # Merge English text
    english_text = " ".join(
        patient_df["text_english"].dropna().astype(str).str.strip()
    )

    results.append({
        "subject_id": subject_id,
        "german_text": german_text,
        "english_text": english_text
    })

# Create output dataframe
result_df = pd.DataFrame(
    results,
    columns=["subject_id", "german_text", "english_text"]
)

# Save as Excel
result_df.to_excel(output_file, index=False)

print(f"Saved {len(result_df)} subjects to: {output_file}")

display(result_df.head())
```
